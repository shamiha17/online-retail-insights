```python

df = pd.get_dummies(df, columns=['Country'], drop_first=True)
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['Day'] = df['InvoiceDate'].dt.day
df['Hour'] = df['InvoiceDate'].dt.hour

df = df.drop('InvoiceDate', axis=1)
```

```python
x=df.drop(['Quantity',	'InvoiceNo'	,'StockCode',	'Description'],axis=1)
y=df['Quantity']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y)
```

```python
x_train.isna().sum()
```

```python
x_train.isna().sum()
```

```python
x_train['CustomerID']=x_train['CustomerID'].fillna(0)
x_test['CustomerID']=x_test['CustomerID'].fillna(0)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from xgboost import XGBRegressor
```

```python
models = {
    "Linear Regression": LinearRegression(),
    
    "Decision Tree": DecisionTreeRegressor(max_depth=5),
    "Random Forest": RandomForestRegressor(n_estimators=100),
    "XGB": XGBRegressor(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=5,
    random_state=42
)
    
}

```

```python
for model_name in models.keys():
    models[model_name].fit(x_train,y_train)
    print(model_name)
    print(models[model_name].score(x_test,y_test))
```

```python

df_sub= pd.DataFrame()
df_sub['CustomerID']=df['CustomerID']
df_sub['Quantity']= models[model_name].predict(x)
df_sub.to_csv("sub.csv",index=False)
```
